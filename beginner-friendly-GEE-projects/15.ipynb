- 🌊 Flood risk mapping
- 💧 Watershed analysis
- 🏔 Mountain studies
- 🚧 Infrastructure planning
- 🌱 Habitat suitability

# ⭐ New Skill: Working with DEM Data

Previously, you worked with satellite imagery:

```text
Satellite Image
      ↓
Calculate NDVI
      ↓
Vegetation Map
```

or

```text
Landsat Thermal Band
        ↓
Temperature Map
```

Today, you'll work with a **Digital Elevation Model (DEM)**:

```text
DEM
 ↓
Elevation
 ↓
Terrain Map
```

Unlike satellite images, a DEM does **not** show color, vegetation, or land cover.

Instead, each pixel stores:

> **The elevation above mean sea level.**

---

# 🏔 What is a DEM?

A **Digital Elevation Model (DEM)** is a raster dataset where every pixel represents **elevation**.

Example pixel values:

```text
150
220
310
480
```

These values mean:

- 150 meters
- 220 meters
- 310 meters
- 480 meters

Unlike NDVI values:

```text
0.65
0.42
```

DEM values are measured in:

> **Meters**

# Step 2 — Load the DEM Dataset

We'll use the **SRTM Digital Elevation Model**.

```javascript
var dem = ee.Image(
  "USGS/SRTMGL1_003"
).clip(countryArea);
```

### What's different?

Previously you loaded datasets like this:

```javascript
ee.ImageCollection(...)
```

Today you use:

```javascript
ee.Image(...)
```

Why?

Because the **SRTM DEM** is provided as **one image**, not an image collection.

# ✅ Complete Code

```javascript
// 1. Study Area
var countryArea = ee.Geometry.Polygon([
  [
    [5.87,47.27],
    [15.04,47.27],
    [15.04,55.06],
    [5.87,55.06],
    [5.87,47.27]
  ]
]);

// 2. Load DEM
var dem = ee.Image(
  "USGS/SRTMGL1_003"
).clip(countryArea);

// 3. Visualization
var demVis = {
  min: 0,
  max: 3000,
  palette: [
    'blue',
    'green',
    'yellow',
    'brown',
    'white'
  ]
};

// 4. Display
Map.centerObject(countryArea,5);

Map.addLayer(
  dem,
  demVis,
  'Elevation'
);
```

# 💡 Things to Remember

## 1. What is a DEM?

A DEM is a raster where every pixel stores:

> **Elevation (meters)**

It describes the **shape of the Earth's surface**, not land cover.

---

## 2. `ee.Image()` vs `ee.ImageCollection()`

Most Sentinel-2 and Landsat datasets are stored as:

```javascript
ee.ImageCollection
```

The SRTM DEM is stored as:

```javascript
ee.Image
```

So you **do not** need:

- `.filterDate()`
- `.sort()`
- `.first()`

---

## 3. Elevation Values

Typical elevation examples:

| Elevation | Description |
|-----------|-------------|
| 0 m | Sea level |
| 250 m | Plain |
| 1200 m | Mountain |
| 3000+ m | High mountains |

---

# ⚠️ Common Beginner Mistakes

## ❌ Mistake 1: Using `.filterDate()` on a DEM

A DEM does not change daily like satellite imagery.

There are **no dates** to filter.

---

## ❌ Mistake 2: Using a visualization range that's too small

For example:

```javascript
min: 0,
max: 100
```

Most mountains would appear with the same color because the visualization range is too limited.

Choose a range that matches the elevation of your study area.

---

## ❌ Mistake 3: Confusing elevation with slope

Remember:

**Elevation** tells you **how high** a location is.

**Slope** tells you **how steep** the land is.

These are different terrain properties.
