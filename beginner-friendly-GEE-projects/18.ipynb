# 🌍 Project 18 — Create an Annual Composite

## 🎯 Goal

Combine all **Sentinel-2 images from one year** into a **single annual composite**.

Instead of displaying one satellite image, we'll create a new image that represents the entire year.

---

## 🌎 Why is this useful in Remote Sensing?

Annual composites are commonly used for:

- 🌍 Land cover mapping
- 🌿 Vegetation studies
- 🏙 Urban monitoring
- 🌾 Agricultural analysis
- 🗺 Creating base maps

Because they combine many images, annual composites greatly reduce the effects of:

- ☁ Clouds
- 🌫 Atmospheric noise
- 📸 Differences between individual image dates

# 📖 What is the Median?

Suppose one pixel has these values during the year:

```
120
135
140
145
900
```

The value **900** might come from a cloud or another unusual observation.

After sorting:

```
120
135
140
145
900
```

The middle value is:

```
140
```

That is the **median**.

Unlike the average (mean), the median is much less affected by extreme values, making it ideal for creating clean composite images.

# Step 1 — Define the Study Area

We'll use Germany again.

```javascript
var countryArea = ee.Geometry.Polygon([
  [
    [5.87,47.27],
    [15.04,47.27],
    [15.04,55.06],
    [5.87,55.06],
    [5.87,47.27]
  ]
]);
```

---

# Step 2 — Load the Sentinel-2 Collection

```javascript
var collection = ee.ImageCollection(
  "COPERNICUS/S2_SR_HARMONIZED"
)
.filterBounds(countryArea)
.filterDate('2024-01-01','2024-12-31');
```

Notice again:

We keep **all images**.

We do **not** use:

```javascript
.first()
```

because we want to combine every image from the year.

---

# Step 3 — Create the Annual Composite ⭐

Today's new function is very simple:

```javascript
var annualComposite = collection.median();
```

Let's understand what happens.

Suppose one pixel has these values in five different images:

| Image | Pixel Value |
|--------|------------:|
| 1 | 180 |
| 2 | 190 |
| 3 | 185 |
| 4 | 700 |
| 5 | 188 |

Earth Engine sorts them:

```
180
185
188
190
700
```

Then selects the middle value:

```
188
```

It repeats this process for **every pixel** and **every band**.

Workflow:

```text
Many Images
      ↓
Median Pixel
      ↓
One Composite Image
```

The result is an:

```text
ee.Image
```

---

# Step 4 — Clip the Composite

```javascript
annualComposite = annualComposite.clip(countryArea);
```

Although every image was filtered by the study area, clipping removes data outside the exact boundary and produces a cleaner display.

---

# Step 5 — Visualization

```javascript
var trueColor = {
  bands:['B4','B3','B2'],
  min:0,
  max:3000
};
```

We'll display the composite in **true color**.

---

# Step 6 — Display

```javascript
Map.centerObject(countryArea,5);

Map.addLayer(
  annualComposite,
  trueColor,
  'Annual Composite'
);
```

The image should look much cleaner than a single satellite image because it combines observations from the entire year.

# 💻 Complete Code

```javascript
// =====================================
// 1. Study Area
// =====================================
var countryArea = ee.Geometry.Polygon([
  [
    [5.87,47.27],
    [15.04,47.27],
    [15.04,55.06],
    [5.87,55.06],
    [5.87,47.27]
  ]
]);

// =====================================
// 2. Load Sentinel-2 Collection
// =====================================
var collection = ee.ImageCollection(
  "COPERNICUS/S2_SR_HARMONIZED"
)
.filterBounds(countryArea)
.filterDate('2024-01-01','2024-12-31');

// =====================================
// 3. Create Annual Composite
// =====================================
var annualComposite = collection.median();

// =====================================
// 4. Clip to Study Area
// =====================================
annualComposite = annualComposite.clip(countryArea);

// =====================================
// 5. Visualization Parameters
// =====================================
var trueColor = {
  bands:['B4','B3','B2'],
  min:0,
  max:3000
};

// =====================================
// 6. Display
// =====================================
Map.centerObject(countryArea,5);

Map.addLayer(
  annualComposite,
  trueColor,
  'Annual Composite'
);
```

# 📌 Things to Remember

## 1. `.median()`

`.median()` combines an **ImageCollection** into a single image by calculating the **median value** of every pixel.

```text
Image Collection
       ↓
median()
       ↓
One Image
```

---

## 2. `qualityMosaic()` vs `.median()`

| `qualityMosaic('NDVI')` | `.median()` |
|--------------------------|-------------|
| Selects the pixel with the highest NDVI | Calculates the median pixel value |
| Best for vegetation studies | Best for creating general cloud-reduced composites |

Both return an **ee.Image**, but they use different methods.

---

## 3. Annual Composite

An annual composite is **not** a real satellite image captured on one day.

Instead, it is a **new image** created by combining many satellite images collected throughout the year.

Different pixels may come from different dates.

---

# ⚠️ Common Beginner Mistakes

### ❌ Mistake 1: Using `.first()`

```javascript
collection.first()
```

This selects only one image, so no composite is created.

---

### ❌ Mistake 2: Expecting the composite to represent one specific date

An annual composite combines observations from many dates.

Different pixels may represent different times during the year.

---

### ❌ Mistake 3: Confusing `.median()` with `qualityMosaic()`

Remember:

```javascript
.median()
```

calculates the **median pixel value**.

Whereas:

```javascript
qualityMosaic('NDVI')
```

selects pixels from the image with the **highest NDVI**.

They solve different problems.
