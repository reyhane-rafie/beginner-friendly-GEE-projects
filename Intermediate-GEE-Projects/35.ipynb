# 🏔️ Project 35 — Calculate Aspect

## 1. Introduce the Project

In the previous project, we calculated **slope**.

Slope tells us:

> **How steep is the land?**

Now we will calculate **aspect**.

Aspect tells us:

> **Which direction is the slope facing?**

For example:

- A slope facing **north** → North-facing slope
- A slope facing **south** → South-facing slope
- A slope facing **east** → East-facing slope
- A slope facing **west** → West-facing slope

Aspect is useful in Remote Sensing, hydrology, agriculture, ecology, and environmental studies because the direction a slope faces can affect **sunlight, temperature, evaporation, snow persistence, and vegetation conditions**.

# 3. Main Concept — What Is Aspect?

Aspect is usually measured in **degrees from north**:

```text
                 North
                  0°
                  │
        NW 315°   │   NE 45°
              \   │   /
               \  │  /
                \ │ /
West 270° ────────┼──────── 90° East
                / │ \
               /  │  \
              /   │   \
        SW 225°   │   SE 135°
                  │
                180°
                South
```

So:

| Aspect | Direction |
|---:|---|
| 0° | North |
| 45° | Northeast |
| 90° | East |
| 135° | Southeast |
| 180° | South |
| 225° | Southwest |
| 270° | West |
| 315° | Northwest |
| 360° | North |

### Important

Aspect is **not the same as slope**.

For example:

```text
Slope = 30°
Aspect = 180°
```

means:

> The land slopes **30° steeply** and faces **south**.

---

# 4. Calculate Aspect in GEE

GEE already provides a terrain function:

```javascript
ee.Terrain.aspect()
```

It calculates the aspect from an elevation image.

So the basic idea is:

```text
DEM
 ↓
ee.Terrain.aspect()
 ↓
Aspect image
```

---

# 5. Step-by-Step

## Step 1 — Load the DEM

```javascript
var dem = ee.Image("USGS/SRTMGL1_003");
```

This is our elevation dataset.

Each pixel contains an elevation value.

---

## Step 2 — Define the Study Area

```javascript
var studyArea = geometry;
```

This assumes you have a geometry drawn in the GEE Code Editor.

---

## Step 3 — Clip the DEM

```javascript
var elevation = dem.clip(studyArea);
```

We only want to work with our study area.

---

## Step 4 — Calculate Aspect

```javascript
var aspect = ee.Terrain.aspect(elevation);
```

This calculates the direction that each slope faces.

The output is an image where each pixel contains an **aspect value in degrees**.

For example:

```text
Pixel A → 0°
Pixel B → 90°
Pixel C → 180°
Pixel D → 270°
```

---

## Step 5 — Display the Aspect Map

```javascript
Map.addLayer(
  aspect,
  {
    min: 0,
    max: 360
  },
  "Aspect"
);
```

This lets GEE visualize the different aspect directions.

---

## Step 6 — Center the Map

```javascript
Map.centerObject(studyArea, 10);
```

This zooms the map to our study area.

---

# 6. Complete Code

```javascript
// ==========================================
// Project 35 — Calculate Aspect
// ==========================================

// 1. Define study area
var studyArea = geometry;

// 2. Load SRTM DEM
var dem = ee.Image("USGS/SRTMGL1_003");

// 3. Clip DEM to study area
var elevation = dem.clip(studyArea);

// 4. Calculate aspect
var aspect = ee.Terrain.aspect(elevation);

// 5. Display elevation
Map.addLayer(
  elevation,
  {
    min: 0,
    max: 3000
  },
  "Elevation"
);

// 6. Display aspect
Map.addLayer(
  aspect,
  {
    min: 0,
    max: 360
  },
  "Aspect"
);

// 7. Center map
Map.centerObject(studyArea, 10);
```

# 9. Why Does Aspect Matter?

Aspect can influence environmental conditions.

### ☀️ Solar radiation

A slope facing toward the sun generally receives different amounts of solar radiation than a slope facing away from it.

### 🌡️ Temperature

Different solar exposure can contribute to differences in surface temperature.

### 💧 Water

Aspect can influence:

- evaporation
- soil moisture
- snowmelt
- runoff patterns

### 🌱 Vegetation

Different slope orientations can create different environmental conditions for vegetation.

For example:

```text
Mountain
   /\
  /  \
 /    \
N      S
```

The two sides can have different sunlight exposure and therefore potentially different vegetation conditions.

**Important:** Aspect alone does not determine these conditions. Latitude, season, time of day, climate, slope, and atmospheric conditions also matter.

# 12. Common Beginner Mistakes

### ❌ Mistake 1 — Thinking 180° means a 180° slope

No.

```text
Aspect = 180°
```

means:

> The slope faces **south**.

It does **not** mean the slope is 180° steep.

---

### ❌ Mistake 2 — Confusing aspect with elevation

Elevation tells you:

> How high?

Aspect tells you:

> Which direction?

---

### ❌ Mistake 3 — Assuming aspect alone determines vegetation

Aspect can influence vegetation conditions, but it is only one environmental factor.

---

### ❌ Mistake 4 — Forgetting that aspect is circular

These are essentially the same direction:

```text
0° = North
360° = North
```

So treating aspect like an ordinary variable from 0 to 360 can sometimes cause problems in statistical/ML analysis.
