## 1. Introduce the Project

In the previous two projects, we learned:

- **Elevation** → How high is the land?
- **Slope** → How steep is the land?
- **Aspect** → Which direction does the land face?

Now we will combine the terrain information to create a **hillshade**.

A hillshade makes a DEM look more like a **3D landscape** by simulating sunlight shining on the terrain.

For example:

```text
        ☀️
         \
          \  light
           \
        /\       /\
       /  \_____/  \
      /             \
```

Areas facing the simulated light become brighter, while areas facing away become darker.

# 3. Main Concept — What Is Hillshade?

**Hillshade** is a visualization of terrain that simulates how the surface would look under a particular light source.

It uses information about:

- **Slope**
- **Aspect**
- **Light direction**
- **Light angle**

The result is a grayscale-like image that gives the terrain a **shaded, 3D appearance**.

### Very important:

Hillshade does **not** change the elevation data.

It is mainly a **visualization product**.

Think of it like putting a virtual flashlight or Sun over your DEM.

---

# 4. Hillshade in GEE

GEE provides:

```javascript
ee.Terrain.hillshade()
```

The basic structure is:

```javascript
var hillshade = ee.Terrain.hillshade(elevation);
```

That's very similar to what we did before:

```javascript
var slope = ee.Terrain.slope(elevation);

var aspect = ee.Terrain.aspect(elevation);

var hillshade = ee.Terrain.hillshade(elevation);
```

Notice the pattern:

```text
DEM
 │
 ├──→ slope
 │
 ├──→ aspect
 │
 └──→ hillshade
```

# 6. Complete Code

```javascript
// ==========================================
// Project 36 — Hillshade Visualization
// ==========================================

// 1. Define study area
var studyArea = ee.Geometry.Polygon([
  [
    [7.55, 47.70],
    [8.35, 47.70],
    [8.35, 48.35],
    [7.55, 48.35],
    [7.55, 47.70]
  ]
]);

// 2. Load DEM
var dem = ee.Image("USGS/SRTMGL1_003");

// 3. Clip DEM
var elevation = dem.clip(studyArea);

// 4. Calculate hillshade
var hillshade = ee.Terrain.hillshade(elevation);

// 5. Display elevation
Map.addLayer(
  elevation,
  {
    min: 200,
    max: 1500
  },
  "Elevation"
);

// 6. Display hillshade
Map.addLayer(
  hillshade,
  {
    min: 0,
    max: 255
  },
  "Hillshade"
);

// 7. Center map
Map.centerObject(studyArea, 10);
```

# 7. Hillshade + Elevation

Here's where hillshade becomes especially useful.

Instead of viewing only the hillshade:

```javascript
Map.addLayer(hillshade, {...}, "Hillshade");
```

you can place it **over the elevation map**.

For example:

```javascript
Map.addLayer(
  elevation,
  {
    min: 200,
    max: 1500
  },
  "Elevation"
);

Map.addLayer(
  hillshade,
  {
    min: 0,
    max: 255,
    opacity: 0.5
  },
  "Hillshade"
);
```

The idea is:

```text
Elevation
   +
Hillshade
   ↓
More realistic terrain visualization
```

`opacity: 0.5` makes the hillshade partially transparent so you can see the elevation underneath it.

# 8. Why Does Hillshade Look 3D?

Imagine a mountain.

If virtual sunlight comes from the northwest:

```text
             ☀️
              \
               \
             /\
            /  \
           /    \
          /      \
```

One side receives the simulated light:

```text
☀️ → → → → → → bright
```

The opposite side becomes darker:

```text
dark ← ← ← ← ← light
```

Our brain interprets these brightness differences as **terrain shape**.

That's why a 2D DEM can suddenly look three-dimensional.

# 10. Things to Remember

### 1. Hillshade simulates illumination

It creates the appearance of sunlight/shadow on terrain.

### 2. It is mainly for visualization

It makes terrain easier to interpret visually.

### 3. It comes from a DEM

```javascript
ee.Terrain.hillshade(elevation)
```

### 4. Hillshade is not slope

```text
Slope     → numerical steepness
Aspect    → numerical direction
Hillshade → visual terrain shading
```

### 5. Hillshade can be combined with other layers

For example:

```text
Elevation
   +
Hillshade
   +
Satellite imagery
```

can produce a very useful terrain visualization.

# 11. Common Beginner Mistakes

### ❌ Mistake 1 — Thinking dark areas are always lower

Not necessarily.

A dark area may simply be a slope facing **away from the simulated light**.

So:

> Dark ≠ low elevation.

---

### ❌ Mistake 2 — Using hillshade as an elevation dataset

Hillshade does not tell you the actual elevation.

Use the DEM for elevation.

---

### ❌ Mistake 3 — Forgetting opacity

If you overlay hillshade at full opacity, it can completely hide your elevation or other layer.

Try:

```javascript
opacity: 0.5
```
