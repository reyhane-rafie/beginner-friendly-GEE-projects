# 🌍 Project 34 — Calculate Slope

## 1. Introduce the Project

### Goal

In this project, we will use a **DEM (Digital Elevation Model)** to create a **slope map**.

A DEM tells us:

> **How high is each location?**

Slope tells us:

> **How steep is each location?**

For example:

```text
Flat land          Steep mountain

──────────           /|
                    / |
                   /  |
                  /   |
````

The final output will be a map showing areas from **flat → steep**.

### Why is slope useful?

Slope is important in:

* erosion studies
* landslide analysis
* watershed studies
* agriculture
* road planning
* hydrology
* habitat analysis

### Main new skills

* Loading a DEM
* Terrain products
* `ee.Terrain.slope()`
* Understanding slope in degrees

---

# 2. Big Picture

```text
DEM
 ↓
Elevation for Every Pixel
 ↓
Calculate Terrain Slope
 ↓
Slope Map
 ↓
Interpret Flat / Steep Areas
```

The important idea is:

```text
DEM → Elevation

Slope → Change in elevation
        across the landscape
```

---

# 3. Main Concept — What Is Slope?

Imagine walking across the landscape.

If elevation changes very little:

```text
──────────────
```

the slope is low.

If elevation changes quickly:

```text
        /
       /
      /
_____/
```

the slope is high.

GEE calculates this from the elevation values surrounding each pixel.

---

# 4. Step 1 — Load a DEM

We will use the **SRTM DEM**.

```javascript
var dem = ee.Image("USGS/SRTMGL1_003");
```

This is an elevation dataset.

Each pixel contains an elevation value, approximately in meters above sea level.

So conceptually:

```text
Pixel A → 125 m
Pixel B → 130 m
Pixel C → 145 m
```

The DEM gives us the terrain height.

---

# 5. Step 2 — Define the Study Area

Use your existing geometry:

```javascript
var studyArea = geometry;
```

Then clip the DEM:

```javascript
var elevation = dem.clip(studyArea);
```

Now we are working only with our study area.

---

# 6. Step 3 — Display the Elevation

```javascript
Map.centerObject(
  studyArea,
  10
);

Map.addLayer(
  elevation,
  {
    min: 0,
    max: 3000
  },
  "Elevation"
);
```

The `min` and `max` values control the visualization range.

They should ideally be adjusted to the elevation range of your study area.

---

# 7. Step 4 — Calculate Slope

Now we use a GEE terrain function:

```javascript
var slope = ee.Terrain.slope(elevation);
```

This is the main new line.

### What does it do?

```text
DEM
 ↓
ee.Terrain.slope()
 ↓
Slope Image
```

The output is another `ee.Image`.

But instead of containing elevation, its pixels contain **slope values in degrees**.

For example:

```text
Pixel A → 2°
Pixel B → 15°
Pixel C → 35°
```

So:

```text
2°  → relatively flat
15° → moderately steep
35° → steep
```

---

# 8. Step 5 — Display the Slope Map

```javascript
Map.addLayer(
  slope,
  {
    min: 0,
    max: 45
  },
  "Slope (degrees)"
);
```

Now we have a slope map.

Conceptually:

```text
0° ─────────────────→ 45°
Flat                  Steep
```

---

# 9. What Does "Degrees" Mean?

This is important.

Slope can be represented in different ways.

We are using **degrees**.

For example:

```text
0°   → completely flat
10°  → gentle slope
30°  → steep slope
45°  → very steep
90°  → vertical
```

These are general interpretations, not universal land-management classifications.

---

# 10. Connect to Previous Lessons

You learned about DEMs in **Project 15 — Elevation Map**.

Previously:

```text
DEM
 ↓
Elevation Map
```

Now we use the same DEM for terrain analysis:

```text
DEM
 ↓
Terrain calculation
 ↓
Slope Map
```

So the important progression is:

```text
Elevation
    ↓
Slope
    ↓
Later:
Aspect
    ↓
Terrain analysis
```

The DEM is therefore not just something we visualize. It can be used to derive other terrain properties.

---

# 11. A Simple Example

Imagine three neighboring pixels:

```text
100 m
   ↓
105 m
   ↓
110 m
```

Elevation is gradually increasing.

The terrain has a relatively gentle slope.

Now imagine:

```text
100 m
   ↓
150 m
   ↓
200 m
```

Elevation changes much faster.

The terrain is much steeper.

GEE performs this type of calculation across the entire DEM.

---

# 12. Complete Workflow

```text
SRTM DEM
   ↓
Elevation Image
   ↓
ee.Terrain.slope()
   ↓
Slope in Degrees
   ↓
Slope Map
```

---

# Complete Code

```javascript
// ========================================
// 1. Study Area
// ========================================

var studyArea = geometry;


// ========================================
// 2. Load DEM
// ========================================

var dem = ee.Image(
  "USGS/SRTMGL1_003"
);


// ========================================
// 3. Clip DEM
// ========================================

var elevation = dem.clip(
  studyArea
);


// ========================================
// 4. Display Elevation
// ========================================

Map.centerObject(
  studyArea,
  10
);

Map.addLayer(
  elevation,
  {
    min: 0,
    max: 3000
  },
  "Elevation"
);


// ========================================
// 5. Calculate Slope
// ========================================

var slope = ee.Terrain.slope(
  elevation
);


// ========================================
// 6. Display Slope
// ========================================

Map.addLayer(
  slope,
  {
    min: 0,
    max: 45
  },
  "Slope (degrees)"
);
```

---

# Things to Remember

### 1. DEM gives elevation

```text
DEM
 ↓
"How high is this location?"
```

### 2. Slope gives steepness

```text
Slope
 ↓
"How quickly does elevation change?"
```

### 3. `ee.Terrain.slope()` creates a slope image

```javascript
var slope = ee.Terrain.slope(dem);
```

The result is an `ee.Image`.

### 4. Slope is expressed in degrees here

```text
0° → flat
higher angle → steeper
```

### 5. Slope is calculated from neighboring elevation values

It isn't simply the elevation value itself.

For example:

```text
Elevation = 2,000 m
```

does **not** mean:

```text
Slope = high
```

A mountain plateau can be at 2,000 m but still have a very low slope.

---

# Common Beginner Mistakes

## ❌ Mistake 1 — Confusing elevation and slope

High elevation does not necessarily mean steep terrain.

```text
High elevation + flat terrain
```

is possible.

And:

```text
Low elevation + steep terrain
```

is also possible.

---

## ❌ Mistake 2 — Forgetting that slope is calculated from the DEM

You cannot calculate meaningful terrain slope from an ordinary satellite RGB image.

You need an elevation surface such as a DEM.

---

## ❌ Mistake 3 — Assuming the visualization range is the actual slope range

If you use:

```javascript
max: 45
```

you are mainly controlling how the map is displayed.

It does **not** mean that the study area cannot contain slopes above 45°.

---

## ❌ Mistake 4 — Assuming one slope classification works everywhere

For example, calling:

```text
0–10° = flat
10–20° = moderate
20–30° = steep
```

may be useful for an exercise, but real research should use classifications appropriate to the specific application.

---

# Your Practice — Project 34.1

Create a **Slope Map** for a mountainous or hilly study area.

### Requirements

1. Define a study area.
2. Load the SRTM DEM.
3. Clip the DEM.
4. Display the elevation.
5. Calculate slope using:

```javascript
ee.Terrain.slope()
```

6. Display the slope map.
7. Compare the elevation map and slope map.

### Your final workflow should be:

```text
DEM
 ↓
Elevation Map

DEM
 ↓
Slope Calculation
 ↓
Slope Map
```

### Extra challenge

Find one area with:

* relatively high elevation but low slope
* relatively low elevation but high slope

This will help you understand the difference between **elevation** and **terrain steepness**.

---

# Conceptual Questions

### 1.

Can an area have **high elevation but low slope**? Why?

### 2.

What type of data does `ee.Terrain.slope()` need as its input?

### 3.

What is the difference between:

```text
Elevation = 2,000 m
```

and:

```text
Slope = 20°
```

?
