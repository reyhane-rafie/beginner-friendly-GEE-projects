# 🛰️ Meet the Three Satellites

## 🛰️ Sentinel-2

- **Spatial Resolution:** 10 m (RGB and NIR)
- **Revisit Time:** ~5 days
- Very detailed imagery
- Excellent for vegetation and agriculture

---

## 🛰️ Landsat 8

- **Spatial Resolution:** 30 m
- **Revisit Time:** 16 days
- Launched in **2013**
- One of the most widely used Earth observation satellites

---

## 🛰️ Landsat 9

- **Spatial Resolution:** 30 m
- **Revisit Time:** 16 days
- Launched in **2021**
- Continues the Landsat mission with improved calibration

---

# Step 1 — Define the Study Area

We'll use **Berlin, Germany**.

```javascript
var studyArea = ee.Geometry.Polygon([
  [
    [13.20,52.35],
    [13.65,52.35],
    [13.65,52.65],
    [13.20,52.65],
    [13.20,52.35]
  ]
]);
```

---

# Step 2 — Load a Sentinel-2 Image

```javascript
var sentinel2 = ee.ImageCollection(
  "COPERNICUS/S2_SR_HARMONIZED"
)
.filterBounds(studyArea)
.filterDate('2024-06-01','2024-06-30')
.sort('CLOUDY_PIXEL_PERCENTAGE')
.first()
.clip(studyArea);
```

---

# Step 3 — Load a Landsat 8 Image

```javascript
var landsat8 = ee.ImageCollection(
  "LANDSAT/LC08/C02/T1_L2"
)
.filterBounds(studyArea)
.filterDate('2024-06-01','2024-06-30')
.sort('CLOUD_COVER')
.first()
.clip(studyArea);
```

### Notice

Sentinel-2 uses:

```javascript
'CLOUDY_PIXEL_PERCENTAGE'
```

Landsat uses:

```javascript
'CLOUD_COVER'
```

Different datasets often have different metadata fields.

---

# Step 4 — Load a Landsat 9 Image

```javascript
var landsat9 = ee.ImageCollection(
  "LANDSAT/LC09/C02/T1_L2"
)
.filterBounds(studyArea)
.filterDate('2024-06-01','2024-06-30')
.sort('CLOUD_COVER')
.first()
.clip(studyArea);
```

---

# Step 5 — Visualization

### Sentinel-2 Visualization

```javascript
var sentinelVis = {
  bands:['B4','B3','B2'],
  min:0,
  max:3000
};
```

### Landsat Visualization

Landsat Collection 2 Level-2 uses different band names.

```javascript
var landsatVis = {
  bands:['SR_B4','SR_B3','SR_B2'],
  min:7000,
  max:18000
};
```

### Notice

Landsat band names are **not** the same as Sentinel-2.

---

# Step 6 — Display the Images

```javascript
Map.centerObject(studyArea,10);

Map.addLayer(
  sentinel2,
  sentinelVis,
  'Sentinel-2'
);

Map.addLayer(
  landsat8,
  landsatVis,
  'Landsat 8'
);

Map.addLayer(
  landsat9,
  landsatVis,
  'Landsat 9'
);
```

Turn the layers on and off.

Ask yourself:

- Which image looks sharper?
- Can you see smaller roads and buildings?
- Do Landsat 8 and Landsat 9 look similar?

# 💻 Complete Code

```javascript
// 1. Study Area
var studyArea = ee.Geometry.Polygon([
  [
    [13.20,52.35],
    [13.65,52.35],
    [13.65,52.65],
    [13.20,52.65],
    [13.20,52.35]
  ]
]);

// 2. Sentinel-2
var sentinel2 = ee.ImageCollection(
  "COPERNICUS/S2_SR_HARMONIZED"
)
.filterBounds(studyArea)
.filterDate('2024-06-01','2024-06-30')
.sort('CLOUDY_PIXEL_PERCENTAGE')
.first()
.clip(studyArea);

// 3. Landsat 8
var landsat8 = ee.ImageCollection(
  "LANDSAT/LC08/C02/T1_L2"
)
.filterBounds(studyArea)
.filterDate('2024-06-01','2024-06-30')
.sort('CLOUD_COVER')
.first()
.clip(studyArea);

// 4. Landsat 9
var landsat9 = ee.ImageCollection(
  "LANDSAT/LC09/C02/T1_L2"
)
.filterBounds(studyArea)
.filterDate('2024-06-01','2024-06-30')
.sort('CLOUD_COVER')
.first()
.clip(studyArea);

// 5. Visualization
var sentinelVis = {
  bands:['B4','B3','B2'],
  min:0,
  max:3000
};

var landsatVis = {
  bands:['SR_B4','SR_B3','SR_B2'],
  min:7000,
  max:18000
};

// 6. Display
Map.centerObject(studyArea,10);

Map.addLayer(
  sentinel2,
  sentinelVis,
  'Sentinel-2'
);

Map.addLayer(
  landsat8,
  landsatVis,
  'Landsat 8'
);

Map.addLayer(
  landsat9,
  landsatVis,
  'Landsat 9'
);
```

# 📌 Things to Remember

## 1. Spatial Resolution

Spatial resolution tells us how much ground one pixel represents.

| Sensor | Spatial Resolution |
|----------|-------------------|
| Sentinel-2 | **10 m per pixel** |
| Landsat 8 | **30 m per pixel** |
| Landsat 9 | **30 m per pixel** |

Smaller pixels capture finer details.

---

## 2. Spectral Resolution

Spectral resolution refers to the number and type of bands a satellite records.

Both Sentinel-2 and Landsat include bands for:

- Blue
- Green
- Red
- Near Infrared (NIR)
- Shortwave Infrared (SWIR)

However:

- The **band names differ**
- The **exact wavelength ranges differ**

Always check the dataset documentation before calculating spectral indices.

---

## 3. Choosing the Right Sensor

| Sensor | Best For |
|----------|----------|
| Sentinel-2 | High-detail mapping, agriculture, vegetation |
| Landsat 8 | Long-term historical studies (since 2013) |
| Landsat 9 | Current Landsat observations with improved calibration |

Choose the sensor based on your **research question**, not because one is always "better."

---

# ⚠️ Common Beginner Mistakes

### ❌ Mistake 1: Using Sentinel-2 band names with Landsat

For example:

```javascript
normalizedDifference(['B8','B4'])
```

works for Sentinel-2 but **not** for Landsat because the band names are different.

---

### ❌ Mistake 2: Comparing images from different dates

If the images were captured months apart, seasonal changes may be mistaken for sensor differences.

---

### ❌ Mistake 3: Assuming higher spatial resolution always means better

Sentinel-2 provides finer detail, but Landsat offers a much longer historical record, making it better for many long-term environmental studies.
