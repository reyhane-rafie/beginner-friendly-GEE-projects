# 💧 Project 37 — Watershed Exploration

## 1. Introduce the Project

So far, we've looked at terrain from different perspectives:

* **Elevation** → how high the land is
* **Slope** → how steep the land is
* **Aspect** → which direction it faces
* **Hillshade** → how the terrain visually looks

Now we're going to connect terrain to **water movement**.

The basic idea of hydrology is:

> **Water generally flows downhill.**

Therefore, elevation and terrain shape can help us understand **where water is likely to accumulate and where it is likely to flow**.

# 3. What Is a Watershed?

A **watershed** is an area of land where water drains toward a common outlet.

Imagine rainfall falling on a mountain:

```text
          /\              /\
         /  \            /  \
        /    \          /    \
       /      \________/      \
              ↓
            River
              ↓
            Outlet
```

Rain falling on one side may eventually reach one river.

Rain falling on the other side may flow toward a completely different river.

The boundary separating these drainage areas is called a **watershed divide**.

### Simple definition

> **A watershed is a land area that drains water to a common point or water body.**

---

# 4. Why Is Elevation Important?

Water doesn't normally flow from low elevation to high elevation.

A simplified model is:

```text
High elevation
      ↓
      ↓ water flow
      ↓
Low elevation
      ↓
River
      ↓
Outlet
```

This means a DEM is extremely useful for hydrological analysis.

However, **elevation alone doesn't explicitly show the flow direction**.

We need to analyze the terrain surface.

---

# 5. A Very Important Hydrology Concept — Drainage

Consider this terrain:

```text
       \       /
        \     /
         \   /
          \ /
           V
          River
```

The terrain slopes toward the valley.

Rainwater tends to move toward the lower area.

A drainage network can therefore often be associated with:

* valleys
* low-elevation channels
* converging terrain
* streams and rivers

---

# 6. Step 1 — Define a Study Area

Let's continue using the **Black Forest** from the previous project:

```javascript
var studyArea = ee.Geometry.Polygon([
  [
    [7.55, 47.70],
    [8.35, 47.70],
    [8.35, 48.35],
    [7.55, 48.35],
    [7.55, 47.70]
  ]
]);
```

This gives us a mountainous region where terrain patterns are easier to see.

---

# 7. Step 2 — Load the DEM

```javascript
var dem = ee.Image("USGS/SRTMGL1_003");
```

Each pixel contains an elevation value.

---

# 8. Step 3 — Clip the DEM

```javascript
var elevation = dem.clip(studyArea);
```

Now we're working only within our study area.

---

# 9. Step 4 — Calculate Slope

```javascript
var slope = ee.Terrain.slope(elevation);
```

Why?

Because slope helps us understand the terrain's steepness.

Areas with larger slope values are steeper.

---

# 10. Step 5 — Create Hillshade

```javascript
var hillshade = ee.Terrain.hillshade(elevation);
```

Hillshade makes the terrain shape easier to see.

This is especially useful when visually exploring drainage patterns.

---

# 11. Step 6 — Display the Terrain

First, display the elevation:

```javascript
Map.addLayer(
  elevation,
  {
    min: 200,
    max: 1500
  },
  "Elevation"
);
```

Then hillshade:

```javascript
Map.addLayer(
  hillshade,
  {
    min: 0,
    max: 255
  },
  "Hillshade"
);
```

You can turn layers on and off in the GEE map.

---

# 12. But Where Is the Drainage?

This is an important point.

A DEM by itself doesn't give us a perfect river network.

To calculate **actual flow direction, flow accumulation, and watershed boundaries**, hydrological algorithms are normally used.

A typical hydrological workflow looks like:

```text
DEM
 ↓
Fill sinks
 ↓
Flow direction
 ↓
Flow accumulation
 ↓
Stream extraction
 ↓
Watershed delineation
```

These concepts are important, but implementing the entire workflow from raw DEM data requires more hydrological processing than we've used so far.

So for this project, we'll first learn to **recognize drainage patterns from terrain**.

---

# 13. Understanding Drainage Patterns from Terrain

Look for areas where terrain appears to converge:

```text
\       /
 \     /
  \   /
   \ /
    │
    │
    │
```

The central low area can represent a **valley or drainage path**.

Compare that with a ridge:

```text
      /\
     /  \
    /    \
   /      \
```

Water tends to move **away from the ridge** toward lower terrain.

This leads to an important relationship:

```text
Ridge
  ↓
Water moves away
  ↓
Valley
  ↓
Stream
```

# 15. Complete Code

```javascript
// ==========================================
// Project 37 — Watershed Exploration
// ==========================================

// 1. Define study area
var studyArea = ee.Geometry.Polygon([
  [
    [7.55, 47.70],
    [8.35, 47.70],
    [8.35, 48.35],
    [7.55, 48.35],
    [7.55, 47.70]
  ]
]);

// 2. Load DEM
var dem = ee.Image("USGS/SRTMGL1_003");

// 3. Clip DEM
var elevation = dem.clip(studyArea);

// 4. Calculate slope
var slope = ee.Terrain.slope(elevation);

// 5. Calculate hillshade
var hillshade = ee.Terrain.hillshade(elevation);

// 6. Display elevation
Map.addLayer(
  elevation,
  {
    min: 200,
    max: 1500
  },
  "Elevation"
);

// 7. Display slope
Map.addLayer(
  slope,
  {
    min: 0,
    max: 45
  },
  "Slope"
);

// 8. Display hillshade
Map.addLayer(
  hillshade,
  {
    min: 0,
    max: 255,
    opacity: 0.5
  },
  "Hillshade"
);

// 9. Center map
Map.centerObject(studyArea, 10);
```

# 17. Things to Remember

### 1. Water generally moves downhill

This is the basic connection between terrain and hydrology.

### 2. Watersheds are drainage areas

A watershed collects water and directs it toward a common outlet.

### 3. Ridges and valleys matter

```text
Ridge → water tends to move away

Valley → water tends to concentrate
```

### 4. DEMs are fundamental for hydrology

A DEM provides the terrain surface needed for many hydrological calculations.

### 5. Terrain interpretation isn't the same as hydrological modeling

Looking at a DEM and identifying likely drainage patterns is **exploration**.

A research-grade watershed analysis normally requires additional steps such as:

```text
DEM
 ↓
Hydrological preprocessing
 ↓
Flow direction
 ↓
Flow accumulation
 ↓
Stream network
 ↓
Watershed boundaries
```

We'll treat those as separate concepts rather than pretending that hillshade itself is a drainage model.

---

# 18. Common Beginner Mistakes

### ❌ Mistake 1 — Thinking water always follows the steepest-looking line on a hillshade

Hillshade is only a visualization.

It doesn't calculate water flow.

---

### ❌ Mistake 2 — Assuming every low pixel is a river

A low area may simply be a valley or depression.

A river requires additional evidence.

---

### ❌ Mistake 3 — Confusing watershed with river

A **river** is a drainage channel.

A **watershed** is the **land area that drains toward a common outlet**.

---

### ❌ Mistake 4 — Thinking elevation alone gives flow direction

Two neighboring pixels can have similar elevations, and the actual terrain structure determines how water moves.

Hydrological algorithms are needed to calculate flow direction and accumulation properly.
